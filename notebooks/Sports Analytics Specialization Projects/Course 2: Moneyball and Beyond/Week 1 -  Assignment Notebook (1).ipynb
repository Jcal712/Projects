```python
# Load packages
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load game logs data
Teams = pd.read_excel("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 2 - Moneyball and Beyond/Week 1 Content/Week 1 Assignment/Game logs 1999-2003 (1).xlsx")
```

```python
# Create a dummy variable for home wins (1 or 0) and visitor wins, and create the ‘year’ variable.
Teams['hwin'] = np.where(Teams['home_score'] > Teams['visitor_score'],1,0)
Teams['awin'] = np.where(Teams['home_score'] < Teams['visitor_score'],1,0)
Teams['year'] = Teams['date'].astype(str).str[0:4]
```

```python
# Create a dataframe that sums the statistics you need by season (year) and by home team. Rename the column for the home team as ‘Team’.
Teamshome = Teams.groupby(['home','year'])[['visitor_ab', 'visitor_h', 'visitor_2b', 'visitor_3b', 'visitor_hr', 'visitor_sf',
                                           'visitor_hbp','visitor_bb','home_ab', 'home_h', 'home_2b', 'home_3b',
                                           'home_hr', 'home_sf', 'home_hbp', 'home_bb', 'hwin','awin']].sum()
Teamshome = Teamshome.reset_index()
Teamshome = Teamshome.rename(columns = {"home": "Team"})
Teamshome
```

```python
# Create a dataframe that sums the statistics you need by season (year) and by visiting team. Rename the column for the visiting team as ‘Team’.
Teamsaway = Teams.groupby(['year','visitor'])[['visitor_ab', 'visitor_h', 'visitor_2b', 'visitor_3b',
                                           'visitor_hr', 'visitor_sf', 'visitor_hbp', 'visitor_bb',
                                           'home_ab', 'home_h', 'home_2b', 'home_3b', 'home_hr',
                                           'home_sf', 'home_hbp', 'home_bb','hwin','awin']].sum()

Teamsaway = Teamsaway.reset_index()
Teamsaway = Teamsaway.rename(columns = {"visitor": "Team"})
Teamsaway
```

```python
# Merge the two dataframes on ‘year’ and ‘Team’ and create a variable for the total number of wins in the season for each team
Teams2 = pd.merge(Teamshome, Teamsaway, on =['Team','year'])
pd.set_option('display.max_columns', 50)
display(Teams2)
```

```python
Teams2['wins'] = Teams2['hwin_x']+Teams2['awin_y']
display(Teams2)
```

# II. Calculating Performance Metrics

```python
# Create batting average for and batting average against as the sum the relevant statistics for each team as home team and visiting team.
Teams2['BAFOR'] = (Teams2['home_h_x']+Teams2['visitor_h_y'])/(Teams2['home_ab_x']+Teams2['visitor_ab_y'])
Teams2['BAAGN'] = (Teams2['home_h_y']+Teams2['visitor_h_x'])/(Teams2['home_ab_y']+Teams2['visitor_ab_x'])
```

```python
# Create the win percentage variable for each team by generating a count of games played and dividing total wins by total games played.
TeamsGh = Teams.groupby(['year','home'])['hwin'].count()
TeamsGh = TeamsGh.reset_index()
TeamsGh = TeamsGh.rename(columns = {"home": 'Team'})

TeamsGa = Teams.groupby(['year','visitor'])['awin'].count()
TeamsGa = TeamsGa.reset_index()
TeamsGa = TeamsGa.rename(columns = {"visitor": 'Team'})

TeamsG = pd.merge(TeamsGh, TeamsGa, on =['Team','year'])
TeamsG['Games'] = TeamsG['hwin']+TeamsG['awin']
TeamsG = TeamsG[['year','Team','Games']]
TeamsG
```

```python
Teams2 = pd.merge(Teams2, TeamsG, on =['Team','year'])
Teams2
```

```python
Teams2['wpc'] = Teams2['wins']/Teams2['Games']
Teams2
```

```python
# Using a similar process to part 2), calculate each team’s home and away win percentage separately for each season.
TeamsGth = Teams.groupby(['year','home'])['hwin'].count()
TeamsGth = TeamsGth.reset_index()
TeamsGth = TeamsGth.rename(columns = {"home": 'Team',"hwin":'home_games'})

TeamsGta = Teams.groupby(['year','visitor'])['awin'].count()
TeamsGta = TeamsGta.reset_index()
TeamsGta = TeamsGta.rename(columns = {"visitor": 'Team',"awin":'away_games'})

TeamsGt = pd.merge(TeamsGth, TeamsGta, on =['Team','year'])
TeamsGt
```

```python
# Merge each team’s home and away win percentage for each season into one data frame.
Teams2 = pd.merge(Teams2, TeamsGt, on =['Team','year'])
Teams2
```

```python
Teams2['hwpc'] = Teams2['hwin_x']/Teams2['home_games']
Teams2['awpc'] = Teams2['awin_y']/Teams2['away_games']
Teams2
```

```python
# Calculate WPCT diff defined as home win percentage – away win percentage for every team in each season.
Teams2['WPCT_diff'] = Teams2['hwpc']-Teams2['awpc']
Teams2
```

```python
# Create a variable AVG Diff = Batting Average For – Batting Average Against
Teams2['AVG_Diff'] = Teams2['BAFOR']-Teams2['BAAGN']
```

# III. Running Regressions

```python
# Run an OLS (Ordinary Least Squares) regression of win percentage on batting average for and batting average against
import statsmodels.formula.api as smf
WinBA_lm = smf.ols(formula = 'wpc ~ BAFOR + BAAGN', data=Teams2).fit()
WinBA_lm.summary()
```

We see a R-squared of 0.720 and coefficients from batting average for and against being 3.7863 and -4.1691 respectively. This means that a one point increase in batting average for will increase win percentage by 3.7863 points and a one point increase in batting average against will decrease the win percentage by 4.1691 points.

```python
# Run an OLS (Ordinary Least Squares) regression of win percentage on batting average for and batting average against,constraining the coefficient on each variable to be equal and opposite.
WinBAcon_lm = smf.ols(formula = 'wpc ~ I(BAFOR - BAAGN)', data=Teams2).fit()
WinBAcon_lm.summary()
```

We see a R-squared of 0.718 and coefficients from batting average for and against being 4.0099 and -4.0099 respectively. This means that a one point increase in batting average for will increase win percentage by 4.0099 points and a one point increase in batting average against will decrease the win percentage by 4.0099 points.
