```python
# First we load the packages
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

# I. Building the Dataset: Player Stats

```python
# Read in salary data retrieved from Lahman's Database
Salary = pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 2 - Moneyball and Beyond/Week 2 Content/Week 2 Assignment/Salaries.csv")

# Read in batting
Batting = pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 2 - Moneyball and Beyond/Week 2 Content/Week 2 Assignment/Batting.csv")
pd.set_option('display.max_columns', 50)
```

```python
# Not every salary is recorded, so first drop missing values
Salary = Salary[Salary.salary > 0]
Salary
```

```python
# Calculate natural log of salary
Salary['lnSal'] = np.log(Salary['salary'])
Salary
```

```python
# Rename 'YearID' as 'SalYear'
Salary = Salary.rename(columns = {'yearID':'SalYear'})
Salary
```

```python
# Let's create a copy of the df which we call "Master", which we can then use to merge the information from all the other dfs
Master = Salary
Master
```

```python
# Using '.groupby' to sum across stints
Batting = Batting.groupby(['playerID','yearID']).sum()
Batting.reset_index(inplace=True)
Batting
```

```python
# Subset Batting data to only include seasons 1998-2006 and batters with at least 130 AB

Batting = Batting[(Batting.yearID >= 1998) & (Batting.yearID <= 2006) & (Batting.AB >= 130)].copy()
display(Batting)

# Now our data consists of 3551 rows × 22 columns
```

```python
# Calculate PA, OBP, SLG and BA in Batting data

Batting['PA'] = Batting['AB'] + Batting['BB'] + Batting['HBP'] + Batting['SH'] + Batting['SF']

Batting['OBP'] = (Batting['H'] + Batting['BB'] + Batting['HBP'])/(Batting['AB'] + Batting['BB']
                                                                  + Batting['HBP'] + Batting['SF'])

Batting['SLG'] = ((Batting['H'] - Batting['Doubles'] - Batting['Triples'] - Batting['HR'])
                  + 2*Batting['Doubles'] + 3*Batting['Triples'] + 4*Batting['HR'])/Batting['AB']

Batting['BA'] = Batting['H']/Batting['AB']

display(Batting)
```

```python
# Create another year variable to link previous season stats to next year's salary in batting data

Batting['SalYear'] = Batting['yearID'] + 1
display(Batting)
```

```python
# Merge Batting into our Master dataset
Master = pd.merge(Batting, Master, on=['SalYear', 'playerID'])
display(Master)
```

# II. Building the Dataset: Player Info

```python
# Read in People data retrieved from Lahman's Database
People = pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 2 - Moneyball and Beyond/Week 2 Content/Week 2 Assignment/People.csv")

# Read in player appearance data retrieved from Lahman's Database
Appearances = pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 2 - Moneyball and Beyond/Week 2 Content/Week 2 Assignment/Appearances.csv")
```

```python
# Extract debut year from debut date in player info data
Debut = People[['playerID','debut']].copy()
Debut['debutyr'] = Debut['debut'].astype(str).str[0:4]
Debut = Debut[['playerID','debutyr']]
display(Debut)
```

```python
# Now merge debut year into master data
Master = pd.merge(Master, Debut, on=['playerID'], how = 'left')
display(Master)
```

```python
# Calculating years of experience
Master['Exp'] = Master['yearID'] - Master['debutyr'].astype(int)
display(Master)
```

```python
# Use experience as proxy to classify players contract status as arbitration eligible or free agent
Master['Arb'] = np.where((Master['Exp'] <= 6) & (Master['Exp'] >= 3),1,0)
Master['Free'] = np.where(Master['Exp'] > 6, 1, 0)
display(Master)
```

```python
# Aggregate appearance data by stint using playerID and yearID
Appearances = Appearances.groupby(['playerID','yearID'])[['G_c','G_1b','G_2b','G_3b','G_ss','G_of','G_dh']].sum()
Appearances.reset_index(inplace=True)
display(Appearances)
```

```python
# Find max number of appearances for each position
Appearances['Max_G'] = Appearances[["G_c","G_1b","G_2b","G_3b","G_ss","G_of","G_dh"]].max(axis=1)
display(Appearances)
```

```python
# Create definiton to position classification process
def Position(df):
    if (df['Max_G'] == df['G_c']): return "C"
    elif (df['Max_G'] == df['G_1b']): return "1B"
    elif (df['Max_G'] == df['G_2b']): return "2B"
    elif (df['Max_G'] == df['G_3b']): return "3B"
    elif (df['Max_G'] == df['G_ss']): return "SS"
    elif (df['Max_G'] == df['G_of']): return "OF"
    elif (df['Max_G'] == df['G_dh']): return "DH"

# Apply newly created definition to appearance data as variable 'POS'
Appearances['POS'] = Appearances.apply(Position, axis = 1)
display(Appearances)
```

```python
# Exclude non-position players (designated hitters) and keep necessary variables
Appearances = Appearances[Appearances['Max_G'] > 0]
Appearances = Appearances[['playerID','yearID','POS']]
display(Appearances)
```

```python
# Create a indicator variable for catcher and infielders (2B, 3B, SS) separately
Appearances['Catch'] = np.where(Appearances['POS'] == "C", 1, 0)
Appearances['SecB'] = np.where(Appearances['POS'] == "2B", 1, 0)
Appearances['ThirB'] = np.where(Appearances['POS'] == "3B", 1, 0)
Appearances['SS'] = np.where(Appearances['POS'] == "SS", 1, 0)
display(Appearances)
```

```python
# Merge position data into Master data
Master = pd.merge(Master, Appearances, on=['playerID','yearID'], how = 'left')
display(Master)
```

# III. Running Regressions

```python
#Separate into pre and post Moneyball Years
Pre_MB_Data = Master[(Master.SalYear >= 1999) & (Master.SalYear <= 2003)]
Post_MB_Data = Master[(Master.SalYear >= 2004) & (Master.SalYear <= 2006)]
```

```python
# Regression for labor market valuation of OBP, SLG and BA for seasons pre Moneyball
import statsmodels.formula.api as smf
Pre_MB_All_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Pre_MB_Data).fit()
Pre_MB_All_lm.summary()
```

```python
# Here we just run each regression by taking a subset for the relevant year, but we don't print out the results.

Pre_MB_1999 = Master[(Master.SalYear == 1999)]
Val_1999_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Pre_MB_1999).fit()
Pre_MB_2000 = Master[(Master.SalYear == 2000)]
Val_2000_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Pre_MB_2000).fit()
Pre_MB_2001 = Master[(Master.SalYear == 2001)]
Val_2001_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Pre_MB_2001).fit()
Pre_MB_2002 = Master[(Master.SalYear == 2002)]
Val_2002_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Pre_MB_2002).fit()
Pre_MB_2003 = Master[(Master.SalYear == 2003)]
Val_2003_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Pre_MB_2003).fit()
```

```python
# Make a Table for all the Pre-Moneyball regressions
from statsmodels.iolib.summary2 import summary_col
Header = ['1999-2003','1999','2000','2001','2002','2003']
Pre_Moneyball_Table = summary_col([Pre_MB_All_lm,Val_1999_lm,Val_2000_lm,Val_2001_lm,Val_2002_lm,Val_2003_lm,],\
                      regressor_order=['OBP','SLG','BA','PA','Arb','Free','Catch','SecB','ThirB','SS','Intercept'],stars=True, \
                      float_format="'%.3f'",model_names = Header)
print(Pre_Moneyball_Table)
```

```python
# Regression for labor market valuation of OBP, SLG and BA for seasons pre Moneyball
Post_MB_All_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Post_MB_Data).fit()
Post_MB_All_lm.summary()
```

```python
# Here we just run each regression by taking a subset for the relevant year, but we don't print out the results.

Post_MB_2004 = Master[(Master.SalYear == 2004)]
Val_2004_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Post_MB_2004).fit()
Post_MB_2005 = Master[(Master.SalYear == 2005)]
Val_2005_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Post_MB_2005).fit()
Post_MB_2006 = Master[(Master.SalYear == 2006)]
Val_2006_lm = smf.ols(formula = 'lnSal ~ OBP + SLG + BA + PA + Arb + Free + Catch + SecB + ThirB + SS', data=Post_MB_2006).fit()
```

```python
# Make a Table for all the Post-Moneyball regressions
Header_2 = ['2004-2006','2004','2005','2006']
Post_Moneyball_Table = summary_col([Post_MB_All_lm,Val_2004_lm,Val_2005_lm,Val_2006_lm,],\
                      regressor_order=['OBP','SLG','BA','PA','Arb','Free','Catch','SecB','ThirB','SS','Intercept'],stars=True, \
                      float_format="'%.3f'",model_names = Header_2)
print(Post_Moneyball_Table)
```

From looking at both of these tables we can corroborate the results of Hakes and Sauer as the effect of OBP on salary gets more prevalent after the publishing of Moneyball.  We can see by the increase in coefficients in the regressions portraying how the ability to draw a walk was undervalued prior to the publication of the Moneyball story.
