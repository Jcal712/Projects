```python
import pandas as pd
import numpy as np
```

### Step 1: Data preparation

```python
NBAmod = pd.read_excel("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 3 - Prediction Models with Sports Data/Week 4 Content/Week 4 Assignment/NBA prediction model (Assignment).xlsx")
NBAmod
```

```python
# Define variables for the probabilities of a home win and away win associated with bookmaker odds
NBAmod['hwinprob']= 1/(NBAmod['hwinodds'])/(1/(NBAmod['hwinodds'])+ 1/(NBAmod['hloseodds']))
NBAmod['awinprob']= 1/(NBAmod['hloseodds'])/(1/(NBAmod['hloseodds'])+ 1/(NBAmod['hwinodds']))
NBAmod
```

```python
# Define a dummy variable = 1 if the home team loses, and zero otherwise
NBAmod['hlose'] = np.where(NBAmod['hwin'] == 1, 0 , 1)
NBAmod
```

```python
# Define a variable equal to H if the home team wins and A if the visiting team wins
NBAmod['winner'] = np.where(NBAmod['hwin'] == 1, 'H' , 'A')
NBAmod
```

```python
# Define a variable equal to H if the home team win probability is greater than 0.5 according to the bookmaker odds and A otherwise
NBAmod['winpred'] = np.where(NBAmod['hwinprob'] >= .5,'H','A')
NBAmod
```

```python
# Define a variable equal to the logarithm of the ratio of the home team salaries to the visiting team salaries.
NBAmod['lTSratio'] = np.log(NBAmod['hteamsal']/NBAmod['opposal'])
NBAmod
```

### Step 2: Estimate a logit model of home time wins depending on the log salary ratio, using the data for calendar year 2018 as the “training data”.

```python
# Define a subset for the calendar year 2018 data
NBAmod_2018 = NBAmod[NBAmod['year']==2018]
NBAmod_2018
```

```python
# Import the logistic regression package
from sklearn.linear_model import LogisticRegression
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
# Run the logistic regression of hwin on the log salary ratio
hwin_Lg = 'hwin~lTSratio'
model = smf.glm(formula = hwin_Lg, data = NBAmod_2018, family=sm.families.Binomial())
result = model.fit()
print(result.summary())
```

### Step 3: Define the predicted probabilities and the predicted results, using the entire data set

```python
# Calculate the predicted probability of home win
NBAmod['pred_Home_W'] = 1/(1+np.exp(-(0.4452 + 1.1216*NBAmod['lTSratio'])))
display(NBAmod)
```

```python
# Based on the predicted probability, define the predicted result H as the outcome where the predicted home win probability is greater than 0.5, and A otherwise.
NBAmod['mod_win_pred'] =  np.where(NBAmod['pred_Home_W'] >= .5,'H','A')
NBAmod
```

### Step 4: For games played in 2019, compare the bookmaker probabilities and model probabilities in terms of the mean number of successfully predicted outcomes and the Brier scores.

```python
# Define the subset of games played in calendar year 2019
NBAmod_2019 = NBAmod[NBAmod['year']==2019]
NBAmod_2019
```

```python
# Define a dummy variable equal to 1 when the bookmaker result prediction is correct, and zero otherwise. Define the equivalent variable for the logit model prediction.
NBAmod_2019['BMcorrect'] = np.where(NBAmod_2019['winpred'] == NBAmod_2019['winner'], 1, 0)
NBAmod_2019['Modcorrect'] = np.where(NBAmod_2019['mod_win_pred'] == NBAmod_2019['winner'], 1, 0)
NBAmod_2019
```

```python
# Mean of BMcorrect (Bookmaker prediction success rate)
NBAmod_2019['BMcorrect'].mean()
```

```python
# Mean of Modcorrect (Model prediction success rate)
NBAmod_2019['Modcorrect'].mean()
```

```python
# Define the Brier score for the bookmaker probabilities and the Brier score for the logit model probabilities
NBAmod_2019['BrierBM']= (NBAmod_2019['hwinprob']-NBAmod_2019['hwin'])**2 + (NBAmod_2019['awinprob']-NBAmod_2019['hlose'])**2
NBAmod_2019['BrierMod'] = (NBAmod_2019['pred_Home_W'] - NBAmod_2019['hwin'])**2 + ((1-NBAmod_2019['pred_Home_W'])-NBAmod_2019['hlose'])**2
```

```python
# Mean of BrierBM (Bookmaker Brier score)
NBAmod_2019['BrierBM'].mean()
```

```python
# Mean of BrierModel (Model Brier score)
NBAmod_2019['BrierMod'].mean()
```
