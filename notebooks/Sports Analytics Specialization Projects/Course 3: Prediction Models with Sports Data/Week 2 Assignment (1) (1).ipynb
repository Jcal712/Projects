```python
# install the packages we need

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the data

NBA19 = pd.read_excel("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 3 - Prediction Models with Sports Data/Week 2 Content/Week 2 Assignment/NBA2019odds.xlsx")
NBA19
```

```python
# Calculate the implied win probability based on the bookmaker odds
NBA19['winprob']= 1/(NBA19['winodds'])/(1/(NBA19['winodds'])+ 1/(NBA19['loseodds']))
NBA19
```

```python
# Drop from the data all games for which home = 0
NBA19 = NBA19[NBA19['home'] != 0]
NBA19
```

```python
# What is the correlation between the home team win probability and home team wins across the entire 2018/19 season?
np.corrcoef(NBA19["winprob"], NBA19['win'])
```

Then create the following subsets

a.	One for games where the absolute value of the points difference is less than 9 and another for games where the points difference is greater than 9 (9 is the median points difference in the data)

b.	One for games that went to overtime, and one for games which finished in regulation time

c.	One for games played in calendar year 2018, one for games played in calendar year 2019

d.	One for each month in the data (October to April)

```python
#a
NBA19_Less_Median =NBA19[abs(NBA19['teampts'] - NBA19['oppopts'])< 9]
NBA19_Greater_Median =  NBA19[abs(NBA19['teampts'] - NBA19['oppopts'])> 9]
NBA19_Less_Median
```

```python
# What is the correlation between the home team win probability and home team wins for games where the points difference was less than 9?
np.corrcoef(NBA19_Less_Median["winprob"], NBA19_Less_Median['win'])
```

```python
# What is the correlation between the home team win probability and home team wins for games where the points difference was greater than 9?
np.corrcoef(NBA19_Greater_Median["winprob"], NBA19_Greater_Median['win'])
```

From these correlations we can assume that uncertain games are the ones where the bookmakers odds are most likely to be wrong and the scores are likely to be the closest.  Games with a low point differential between a huge favorite and a huge underdog might not even make it possible for the favorite to cover a game.  These factors are what makes any sports only forecastable to a certain degree as there are random underlying factors that could just pop up on any given night.

```python
#b
NBA19_Overtime = NBA19[NBA19['overtime'] != 0]
NBA19_Regulation =  NBA19[NBA19['overtime'] == 0]
NBA19_Overtime
```

```python
# What is the correlation between the home team win probability and home team wins for games where the game went to overtime?
np.corrcoef(NBA19_Overtime["winprob"], NBA19_Overtime['win'])
```

```python
# What is the correlation between the home team win probability and home team wins for games where the game was finished in regular time?
np.corrcoef(NBA19_Regulation["winprob"], NBA19_Regulation['win'])
```

From these correlations we can assume that uncertain games are the ones where the bookmakers odds are most likely to be wrong and these games are the ones that tend to go into overtime. Overtime games are usually closer, so a big spread that goes into overtime might not even be possible to cover a game once it has gone into overtime making it harder to predict the outcome initially without knowing the game will go into overtime.
