```python
#Import Libraries

import pandas as pd
import numpy as np
import datetime
import matplotlib.pyplot as plt
import seaborn as sns
```

**Step 1:** Import the “NHL_Team.csv” data file and name the dataframe as “NHL_Team” in Jupyter Notebook.

a) Delete the following variables: “Unnamed:0”, “abbr”, “tname”, “lname”, and “sname”.

b) Rename the variable “name” to “team_name”.

```python
#Import NHL Team Data and Display
NHL_Team=pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 1 - Foundations of Sports Analytics: Data, Representation, and Models in Sports/Week 2 Content/Week 2 Assignment/NHL_team.csv")
NHL_Team
```

```python
# Deleting the required variables and checking dataframe
NHL_Team.drop(['Unnamed: 0','abbr','tname','lname','sname'], axis=1, inplace=True)
NHL_Team
```

```python
# Renaming name to team_name and check dataframe
NHL_Team.rename(columns={'name':'team_name'}, inplace=True)
NHL_Team
```

**Step 2:** Import the “NHL_competition.csv” data file and name the dataframe as “NHL_Competition” in Jupyter Notebook.

a) Delete the following variables: “Unnamed: 0”, “tz”, “start”, and “end”

b) Rename the variable “name” to “competition_name”.

```python
#Import NHL Competition Data and Display Head
NHL_Competition=pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 1 - Foundations of Sports Analytics: Data, Representation, and Models in Sports/Week 2 Content/Week 2 Assignment/NHL_competition.csv")
NHL_Competition.head()
```

```python
# Deleting the required variables and checking dataframe
NHL_Competition.drop(['Unnamed: 0','tz','start','end'], axis=1, inplace=True)
NHL_Competition
```

```python
# Renaming name to competition_name and check dataframe
NHL_Competition.rename(columns={'name':'competition_name'}, inplace=True)
NHL_Competition
```

**Step 3:** Import the “NHL_game.csv” data file and name the dataframe as “NHL_Game” in Jupyter Notebook.

a) Delete the following variables: “X”, “period”, and “status”.

```python
#Import NHL Game Data and Display Head
NHL_Game=pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 1 - Foundations of Sports Analytics: Data, Representation, and Models in Sports/Week 2 Content/Week 2 Assignment/NHL_game.csv")
NHL_Game.head()
```

```python
# Deleting the required variables and checking dataframe
NHL_Game.drop(['X','period','status'], axis=1, inplace=True)
NHL_Game
```

**Step 4:** Merge the dataframe “NHL_Team” into the dataframe “NHL_Game” by “tid.” Continue to name the merged dataframe as “NHL_Game.”

```python
# Merge the NHL_Team and NHL_Game dataframes on tid and assign back to NHL_Game df.
NHL_Game = pd.merge(NHL_Team,NHL_Game, on='tid')
NHL_Game
```

**Step 5:** Merge the dataframe “NHL_Competition” into the dataframe “NHL_Game” by “comp_id.” Continue to name the merged dataframe as “NHL_Game.”

```python
# Merge the NHL_Competition and NHL_Game dataframes on comp_id and assign back to NHL_Game df.
NHL_Game = pd.merge(NHL_Competition,NHL_Game, on='comp_id')
NHL_Game
```

**Step 6:** In the merged “NHL_Game” dataframe, create a variable “hgd” to indicate the goal difference between home and away score (hscore – ascore) and delete observations with missing value in the variable “hgd”.

```python
# Create variable hgd in NHL_Game df to indicate goal difference between home and away scores and delete rows where hgd is missing
NHL_Game['hgd'] = NHL_Game['hscore']- NHL_Game['ascore']
NHL_Game=NHL_Game[pd.notnull(NHL_Game["hgd"])]
NHL_Game
```

**Step 7:** Drop all observations with missing values, if there is still any, from the “NHL_Game” dataframe.

```python
# Drop the rest of the rows with missing values from NHL_Game
NHL_Game=NHL_Game.dropna()
```

**Step 8:** Convert the type of the “date” variable from “object” to “datetime.”

```python
# Convert date variable to datetime
NHL_Game['date']=pd.to_datetime(NHL_Game['date'])
NHL_Game
```

**Step 9:** Sort the NHL games by “date” and show the first 15 observations.

```python
# Sort the NHL_Game df by date and show the first 15 entries
NHL_Game = NHL_Game.sort_values(by= 'date',ascending=True)
NHL_Game.head(15)
```

**Step 10:** Create two dataframes that separate the “NHL_Game” dataframe by home and away games. Name them “NHL_Home” and “NHL_Away”, respectively.

a) Rename variables:
     
     i) For away games, rename “ascore” to “goals_for”; rename “hscore” to “goals_against”
     ii) For home games, rename “hscore” to “goals_for”; rename “ascore” to “goals_against”
b) Create a “win” variable that equals to 1 if the team won the game; 0 if the team lost the game; and 0.5 if it was a draw.

```python
# Create a home and away dataframe for home and away games respectively
NHL_Home = NHL_Game[NHL_Game.home_away == 'home'].reset_index()
NHL_Away = NHL_Game[NHL_Game.home_away == 'away'].reset_index()
NHL_Home
```

```python
# Rename the variables in both dataframes to specify home or away
NHL_Home = NHL_Home.rename(columns={'ascore':'goals_against','hscore':'goals_for'})
NHL_Away = NHL_Away.rename(columns={'ascore':'goals_for','hscore':'goals_against'})
NHL_Away.head()
```

```python
# Create a win variable for home and away that equals 1 for a win, 0 for a loss, and 0.5 for a tie
NHL_Home['win']=np.where(NHL_Home['hgd'] > 0,1,np.where(NHL_Home['hgd'] < 0,0,0.5))
NHL_Away['win']=np.where(NHL_Away['hgd'] < 0,1,np.where(NHL_Away['hgd'] > 0,0,0.5))
NHL_Home.head()
```

**Step 11:** Append the “NHL_Home” and “NHL_Away” dataframes to be the new “NHL_Game” dataframe.

```python
NHL_Game = pd.concat([NHL_Home, NHL_Away], ignore_index=True)
NHL_Game
```

**Step 12:** Generate a team level dataframe that aggregates the total number of games won, the total number of “goals_for” and “goals_against” for each team in each competition (i.e. grouped by tid, competition_name and type). Name this new dataframe “NHL_Team_Stats”. Make sure to convert the indexes of the new dataframe back as  variables.

```python
# Create a NHL_Team_Stats df grouped by tid, competition_name, and type and sum the wins, goals_for, and goals_against for each entry
NHL_Team_Stats = NHL_Game.groupby(['tid','competition_name','type'])[['win','goals_for','goals_against']].sum().reset_index()
NHL_Team_Stats
```

**Step 13:** Create a dataframe “NHL_Game_Count” that include the total number of games played by each team in each competition (i.e. grouped by tid, competition_name and type). Name this new variable in the dataframe “game_count”.

```python
# Create a NHL_Game_Count df grouped by tid, competition_name, and type and sum the total number of games played for each entry
NHL_Game_Count = NHL_Game.groupby(['tid','competition_name','type'],as_index=False).size()
NHL_Game_Count = NHL_Game_Count.rename(columns={'size':'game_count'})
NHL_Game_Count
```

**Step 14:** Merge dataframes.

a) Merge the “NHL_Game_Count” dataframe into the “NHL_Team_Stats” dataframe by “tid”, “competition_name”, and “type”. Continue to name the merged dataframe “NHL_Team_Stats”.

b) Merge the “NHL_Team” dataframe into the “NHL_Team_Stats” dataframe by “tid”. Continue to name the merged dataframe “NHL_Team_Stats”.

```python
# Merge NHL_Game_Count into NHL_Team_Stats by tid, competition_name, and type
NHL_Team_Stats = pd.merge(NHL_Team_Stats,NHL_Game_Count, on=['tid','competition_name','type'])
NHL_Team_Stats
```

```python
# Merge NHL_Team into NHL_Team_Stats by tid
NHL_Team_Stats = pd.merge(NHL_Team_Stats,NHL_Team, on=['tid'])
NHL_Team_Stats
```

**Step 15**: Import the “pp.pk.ppgf.csv” data file and name the dataframe as “NHL_PPPK” in Jupyter Notebook. Merge the “NHL_PPPK” dataframe into the “NHL_Team_Stats” dataframe by “tricode” and “competition_name”.

```python
#Import NHL PPPK Data and Display Head
NHL_PPPK=pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 1 - Foundations of Sports Analytics: Data, Representation, and Models in Sports/Week 2 Content/Week 2 Assignment/pp.pk.ppgf.csv")
NHL_PPPK.head()
```

```python
# Merge NHL_PPPK into NHL_Team_Stats by tricode and competition_name
NHL_Team_Stats = pd.merge(NHL_Team_Stats,NHL_PPPK, on=['tricode','competition_name'])
NHL_Team_Stats
```

**Step 16:** Create new variables in the “NHL_Team_Stats” dataframe.

a) Winning percentage (“win_pct”)=”win”/ total number of games played

b) Average goals for per game (“avg_gf”)=total number of goals for / total number of games played     

c) Average goals against per game (“avg_ga”)=total number of goals against / total number of games played

```python
# Create new winning percentage, average goals for per game, and average goals against per game variables in NHL_Team_Stats df
NHL_Team_Stats['win_pct'] = NHL_Team_Stats['win']/NHL_Team_Stats['game_count']
NHL_Team_Stats['avg_gf'] = NHL_Team_Stats['goals_for']/NHL_Team_Stats['game_count']
NHL_Team_Stats['avg_ga'] = NHL_Team_Stats['goals_against']/NHL_Team_Stats['game_count']
NHL_Team_Stats
```

**Step 17:** In the “NHL_Competition” dataframe, the variable “type” indicates the type of competition: type=2 – regular season. Create a dataframe that contains team statistics for games only during regular seasons. Name this dataframe “NHL_Team_R_Stats”.

```python
# Create dataframe that contains the team statistics for regular seasons only
NHL_Team_R_Stats = NHL_Team_Stats[NHL_Team_Stats.type == 2].reset_index()
NHL_Team_R_Stats
```

## Part 3 - Correlation Analyses


**Step 1:** In the “NHL_Team_R_Stats” dataframe, make a scatter plot to depict the relationship between the total number of goals for and the winning percentage.

a) Plot the total number of goals for on the x-axis and winning percentage on the y-axis.

b) Add a regression line to the scatter plot.

c) Make the title of the graph “Relationship between Goals for and Winning Percentage” and make the font size 11.

d) Label the x-axis “Total Goals for” and label the y-axis “Winning Percentage”.

```python
# Create a scatter plot of NHL_Team_R_Stats dataframe showing relationship between total number of goals and winning percentage using required settings.
sns.regplot(x='goals_for', y='win_pct', data=NHL_Team_R_Stats,  marker='.')
plt.xlabel('Total Goals for')
plt.ylabel('Winning Percentage')
plt.title("Relationship between Goals for and Winning Percentage", fontsize=11)
```

**Step 2:** In the “NHL_Team_R_Stats” dataframe, calculate the correlation coefficient between total number of goals for and winning percentage.

```python
# Calculate correlation coefficient for goals_for and win_pct in NHL_Team_R_Stats dataframe
NHL_Team_R_Stats['goals_for'].corr(NHL_Team_R_Stats['win_pct'])
```

**Step 3:** Create a scatter plot of the total number of goals for and winning percentage similar to step 1 for regular season games. In this graph, group observations by “competition_name”.

a) Plot the total number of goals for on the x-axis and winning percentage on the y-axis.

b) Add a regression line to the scatter plot.

c) Make the title of the graph “Relationship between Goals for and Winning Percentage” and make the font size 11.

d) Label the x-axis “Total Goals for” and label the y-axis “Winning Percentage”.

```python
# Create a scatter plot of NHL_Team_R_Stats dataframe showing relationship between total number of goals and winning percentage.
# Group by competition name and use required settings.
sns.lmplot(x='goals_for', y='win_pct', data=NHL_Team_R_Stats, hue='competition_name')
plt.xlabel('Total Goals for')
plt.ylabel('Winning Percentage')
plt.title("Relationship between Goals for and Winning Percentage", fontsize=11)
```

From the above graph we can see that the 2011 and 2012 season are skewing the correlation and regression line.  We can drop those two seasons and compare.

**Step 4:** For the “NHL_Team_R_Stats” dataframe, delete observations of 2011 and 2012 seasons. Continue to name the dataframe “NHL_Team_R_Stats”.

```python
# Delete observations from NHL_Team_R_Stats where competition_name is 2011 and 2012 regular seasons
NHL_Team_R_Stats=NHL_Team_R_Stats[(NHL_Team_R_Stats.competition_name != '2011 NHL Regular Season')&(NHL_Team_R_Stats.competition_name!='2012 NHL Regular Season')]
```

**Step 5:** In the new “NHL_Team_R_Stats” dataframe, create a scatter plot of total number of goals for and winning percentage.

a) Plot the total number of goals for on the x-axis and winning percentage on the y-axis.

b) Add a regression line to the scatter plot.

c) Make the title of the graph “Relationship between Goals for and Winning Percentage” and make the font size 11.

d) Label the x-axis “Total Goals for” and label the y-axis “Winning Percentage”.

```python
# Create a scatter plot of the new NHL_Team_R_Stats dataframe showing relationship between total number of goals and winning percentage using required settings.
sns.regplot(x='goals_for', y='win_pct', data=NHL_Team_R_Stats,  marker='.')
plt.xlabel('Total Goals for')
plt.ylabel('Winning Percentage')
plt.title("Relationship between Goals for and Winning Percentage", fontsize=11)
```

**Step 6:** Calculate the correlation coefficient between total number of goals for and winning percentage in the updated “NHL_Team_R_Stats” dataframe.

```python
# Calculate correlation coefficient for goals_for and win_pct in new NHL_Team_R_Stats dataframe
NHL_Team_R_Stats['goals_for'].corr(NHL_Team_R_Stats['win_pct'])
```

From the graph of the new NHL_Team_R_Stats dataframe, we can see that the regression line can be much more clearly observed.
