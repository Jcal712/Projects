## Part 2 - Regression Analysis 1 - Test of Home Game Advantage

In this regression analysis, you will try to determine if playing at home gives teams any advantage in their performance.

Run the following regressions where the variable “score_diff” is the dependent variable.

```python
# Regression of score_diff on home
Reg1_1 = sm.ols(formula = 'score_diff ~ home', data= NFL_Game).fit()
print(Reg1_1.summary())
```

Our regression tells us that if a team is home the score diff will increase by 5.22.  We also have a high t-value and p-value less than 0.05 indicating that this is statistically significant.  The R-Squared however is only 0.030 indicating that our regression only captures about 3 percent of variation in the data.  We can definitely get a better model which we will try to do with the second regression.

```python
# Regression of score_diff on home, stadium_capacity, stadium_neutral, and the interaction between home and stadium neutral
Reg1_2 = sm.ols(formula = 'score_diff ~ home+stadium_capacity+stadium_neutral+home*stadium_neutral', data= NFL_Game).fit()
print(Reg1_2.summary())
```

Our regression tells us

*   If a team is home the score diff will increase by 5.28.
*   If the stadium capacity increases by one the score diff will increase by 0.0000000052
*   If a game is at a neutral site the score diff will increase by 7.25.
*   If the team is a home team and the game is played at a neutral site the score diff will be 5.2765-14.5038 = -9.2273 meaning it will decrease by 9.2273


We also have high t-value and p-value less than 0.05 for all variables except stadium capacity indicating that all the interactions are statistically significant except for stadium capacity. The R-Squared only increases 0.031 indicating that our regression only captures about 3.1 percent of variation in the data. This is better than our first model by a slim margin but maybe our last model of this type can get us a better interpretation of the data.

```python
# Regression of score_diff on home, stadium_capacity, stadium_neutral, the interaction between home and stadium neutral and the interaction between team and opponent.
Reg1_3 = sm.ols(formula = 'score_diff ~ home+stadium_capacity+stadium_neutral+home*stadium_neutral+team*opponent', data= NFL_Game).fit()
print(Reg1_3.summary())
```

This regression shows us the statistics for the variables explored in the last model but adds the interaction between every team and opponent which gives us a lot more variables.  Looking at the R-squared we can see it increased from the 0.031 we got in our last model to 0.128 meaning this model captures 12.8 percent of variation.  This model is about 4 times better than our last one.  Now we will explore the impact of outside factors on score.

## Part 3 - Regression Analysis 2 -- Impact of Outside Factors on Scores

In this regression analysis, you will investigate if the final score earned by each team is affected by outside factors such as the size and condition of the stadium as well as weather conditions at the stadium.

Run the following regressions where the variable “score” is the dependent variable.

```python
# Regression of score on season and home
Reg2_1 = sm.ols(formula = 'score ~ season+home', data= NFL_Game).fit()
print(Reg2_1.summary())
```

Our regression tells us that if a team is home their score will increase by 2.61 points. Also, as the season increases by one year the score will increase by 0.054 points.  We have a high t-value and p-value less than 0.05 indicating that these relationships are statistically significant. The R-Squared however is only 0.022 indicating that our regression only captures about 2.2 percent of variation in the data. We can definitely get a better model which we will try to do with the second regression.

```python
# Regression of score on season, home, weather_temperature, weather_wind_mph, and weather_humidity
Reg2_2 = sm.ols(formula = 'score ~ season+home+weather_temperature+weather_wind_mph+weather_humidity', data= NFL_Game).fit()
print(Reg2_2.summary())
```

Our regression tells us that

*   If a team is home their score will increase by 2.61 points.
*   As the season increases by one year the score will increase by 0.04 points.
*   As the temperature increases by one degree score will decrease by 0.0006 points.
*   As the wind speed increases by one mph score will decrease by 0.1188 points.
*   As the humidity increases by one percent score will decrease by 0.015 points.

We have a high t-value and p-value less than 0.05 for all variables besides weather_temperature indicating that these relationships are all statistically significant except for temperature. The R-Squared however increases slightly to 0.026 indicating that our regression only captures about 2.6 percent of variation in the data. We will try to get a better model in our third regression.

```python
# Regression of score on season, home, weather_temperature, weather_wind_mph, weather_humidity, stadium_capacity, stadium_age, stadium_type, stadium_neutral, and the interaction between home and stadium_neutral
Reg2_3 = sm.ols(formula = 'score ~ season+home+weather_temperature+weather_wind_mph+weather_humidity+stadium_capacity+stadium_age+stadium_type+stadium_neutral+home*stadium_neutral', data= NFL_Game).fit()
print(Reg2_3.summary())
```

Our regression tells us that

*   If a team is home their score will increase by 2.64 points.
*   As the season increases by one year the score will increase by 0.039 points.
*   As the temperature increases by one degree score will decrease by 0.0009 points.
*   As the wind speed increases by one mph score will decrease by 0.1329 points.
*   As the humidity increases by one percent score will decrease by 0.0144 points.
*   If a team plays in an outdoor stadium their score will increase by 0.326 points.
*   If a team plays in a retractable stadium their score will increase by 0.597 points.
*   If the stadium increases by one person, the team's score will decrease by 0.0000124 points.
*   As the stadium age increases by one year, the team's score will increase by 0.0053 points.
*   If the team plays at a neutral site, their score will increase by 3.85 points
*   If the home team plays at a neutral site, their score decrease by (2.6382-7.2545 = -4.6163) 4.616 points.

We have a high t-value and p-value less than 0.05 for season, home, weather_wind_mph, weather_humidity, stadium_neutral, and home*stadium_neutral indicating that these relationships are all statistically significant where all other variables present are not statistically significant. The R-Squared again increases slightly from 0.26 to 0.27 indicating that our regression only captures about 2.7 percent of variation in the data. We will try one last time to get a better model in our final regression.

```python
# Regression of score on season, home, weather_temperature, weather_wind_mph, weather_humidity, stadium_capacity, stadium_age, stadium_type, stadium_neutral, team, opponent and the interaction between home and stadium_neutral
Reg2_4 = sm.ols(formula = 'score ~ season+home+weather_temperature+weather_wind_mph+weather_humidity+stadium_capacity+stadium_age+stadium_type+stadium_neutral+home*stadium_neutral+team+opponent', data= NFL_Game).fit()
print(Reg2_4.summary())
```

This regression shows us the statistics for the variables explored in the last model but adds the team and opponent which allows us to look at an increase or decrease in score based on which team is playing and who their opponent is. Looking at the R-squared we can see it increased from the 0.027 we got in our last model to 0.052 meaning this model captures 5.2 percent of variation. This model is about 2 times better than our last one.
