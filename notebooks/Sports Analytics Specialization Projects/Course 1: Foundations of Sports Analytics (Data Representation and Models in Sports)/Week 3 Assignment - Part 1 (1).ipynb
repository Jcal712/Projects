```python
# As usual, we begin by importing the packages we will need
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# The data consists of the shot log for the NBA season 2016/17
shot = pd.read_csv("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 1 - Foundations of Sports Analytics: Data, Representation, and Models in Sports/Week 3 Content/Week 3 Assignment: Part 1/NBA Shotlog_16_17 (1).csv")
pd.set_option('display.max_columns', 100)
print(shot.columns.tolist())
shot.describe()
```

```python
# Using code from previous lesson to mirror shots on the left basket to the right basket to get a more complete picture
shot['halfcourt_x'] =np.where(shot['location_x'] < 933/2, 933 - shot['location_x'],shot['location_x'])
shot['halfcourt_y'] =np.where(shot['location_x'] < 933/2, 500 - shot['location_y'],shot['location_y'])
```

## Westbrook - Durant Comparision

```python
# Create subsets for Russell Westbrook and Kevin Durant
Westbrook = shot[shot['shoot_player']=='Russell Westbrook']
Durant = shot[shot['shoot_player']=='Kevin Durant']
```

```python
# Create 2 new columns containing the coordinates of each shot for Westbrook and Durant
hxW = Westbrook['halfcourt_x']
hyW = Westbrook['halfcourt_y']
hxD = Durant['halfcourt_x']
hyD = Durant['halfcourt_y']
```

```python
# Plot the shots for Westbrook and Durant side by side so we can compare
f = plt.figure(figsize=(94/6,50/6))
ax = f.add_subplot(121)
colors = np.where(Westbrook['current_shot_outcome']=='SCORED','r',np.where(Westbrook['current_shot_outcome']=='MISSED','b','g'))
ax = plt.scatter(hxW,hyW, s=10, c= colors, marker= '.')
plt.grid(True)
plt.title("Russell Westbrook", fontsize = 15)
plt.xlim(500,950)
plt.ylim(0,500)

ax = f.add_subplot(122)
colors = np.where(Durant['current_shot_outcome']=='SCORED','r',np.where(Durant['current_shot_outcome']=='MISSED','b','g'))
ax = plt.scatter(hxD,hyD, s=10, c= colors, marker= '.')
plt.grid(True)
plt.title("Kevin Durant", fontsize = 15)
plt.xlim(500,950)
plt.ylim(0,500)
```

These plots tell a pretty similar story.  It seems as both Westbrook and Durant are not afraid to take their shots from anywhere as both players shoot from behind the arc, midrange, and in the paint.  One stark difference between the two is the density of shots.  Westbrook's shots are more dense than Durant's at every spot on the court.  Since this was before I started watching basketball I did some research and found out two factors that would contribute to this.  First thing is Kevin Durant missed 20 games with an MCL sprain during the season while Westbrook only missed 1 game which creates a 19 game disparity worth of shots.  The second thing is that 2016-17 was the year that Russell Westbrook won the MVP which means he was far and away the best player that season and leads to the assumption that he would have taken more shots.

## Jordan - Howard Comparision

```python
# Create subsets for DeAndre Jordan and Dwight Howard
Jordan = shot[shot['shoot_player']=='DeAndre Jordan']
Howard = shot[shot['shoot_player']=='Dwight Howard']
```

```python
# Create 2 new columns containing the coordinates of each shot for Jordan and Howard
hxJ = Jordan['halfcourt_x']
hyJ = Jordan['halfcourt_y']
hxH = Howard['halfcourt_x']
hyH = Howard['halfcourt_y']
```

```python
# Plot the shots for Jordan and Howard side by side so we can compare
f = plt.figure(figsize=(94/6,50/6))
ax = f.add_subplot(121)
colors = np.where(Jordan['current_shot_outcome']=='SCORED','r',np.where(Jordan['current_shot_outcome']=='MISSED','b','g'))
ax = plt.scatter(hxJ,hyJ, s=10, c= colors, marker= '.')
plt.grid(True)
plt.title("DeAndre Jordan", fontsize = 15)
plt.xlim(500,950)
plt.ylim(0,500)

ax = f.add_subplot(122)
colors = np.where(Howard['current_shot_outcome']=='SCORED','r',np.where(Howard['current_shot_outcome']=='MISSED','b','g'))
ax = plt.scatter(hxH,hyH, s=10, c= colors, marker= '.')
plt.grid(True)
plt.title("Dwight Howard", fontsize = 15)
plt.xlim(500,950)
plt.ylim(0,500)
```

These plots tell a pretty similar story as well. Both Jordan and Howard take large majority of their shots deep in the paint. One difference between the two is the midrange. Jordan took five 3-pointers with almost every other shot being deep in the paint with only one a couple being classified as midrange.  Howard on the other hand also only took a handful of threes and takes a majority of his shots from very close to the basket, but, he actually has taken midrange shots which DeAndre Jordan has not.  From this what we can tell is that Howard believed in his midrange shooting more than Jordan did and most of their shots are from inside the paint due to them both being centers which live in the paint.
