```python
# As usual, we begin by loading the packages we will need

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
```

```python
# Now we load the data

IPL=pd.read_excel("/content/drive/MyDrive/Sports Performance Analytics Michigan/Course 1 - Foundations of Sports Analytics: Data, Representation, and Models in Sports/Week 5 Content/Week 5 Assignment/IPL (assignment) data.xlsx")
```

```python
IPL
```

```python
# Create the sum of salaries in each season
Sumsal = IPL.groupby(['year'])['salaries'].sum().reset_index().rename(columns={'salaries':'allsal'})
Sumsal
```

```python
# Merge the two dataframes on year
IPL = pd.merge(IPL, Sumsal, on=['year'], how='left')
display(IPL)
```

```python
# Create a variable for team salary divided by total salaries for that season (relsal).
IPL['relsal']= IPL['salaries']/IPL['allsal']
```

```python
# Creating a value for win percentage. Defined as wins divided by games with a result (= games played minus games with no result).
IPL['win_pct']=IPL['won'] / (IPL['played']-IPL['noresult'])
```

```python
# Taking another look at our dataframe with our newly created variables.
IPL
```

```python
# first we sort the values
IPL.sort_values(by=['team','year'], ascending=True)
```

```python
# now we create the lagged dependend variable
IPL['wpc_lag'] = IPL.groupby('team')['win_pct'].shift(1)
IPL
```

```python
# Regression of win_pct on relsal
lm_1 = smf.ols(formula = 'win_pct ~ relsal', data=IPL).fit()
print(lm_1.summary())
```

From the above regression we can see this model is not very good.  We have a R-squared of 0.002 saying that our regression explains 0.2 percent of variation in our data.  We also see that our only dependent variable of relsal has a p-value way higher than 0.05 meaning that it is not significant in our regression.  Next we will try to add the lagged win percentage to this model to see if we can get a stronger outcome.

```python
# Regression of win_pct on relsal and wpc_lag
lm_2 = smf.ols(formula = 'win_pct ~ relsal+wpc_lag', data=IPL).fit()
print(lm_2.summary())
```

This regression is a little better but still not very good and we fall in to a couple of the same problems.  Firstly, our R-squared is 0.040 meaning our regression exlpains about 4 percent of variation in our data which we would like to be much higher.  Lastly, again, both of our dependent variables have a p-value substantially higher than 0.05 meaning they aare not significant.  Now let's try one final regression adding in the fixed effects for each team.

```python
# Regression of win_pct on relsal, wpc_lag, and team fixed effects
lm_3 = smf.ols(formula = 'win_pct ~ relsal+wpc_lag+C(team)', data=IPL).fit()
print(lm_3.summary())
```

This regression shows big improvements over our first two models.  Our R-squared for this regression sits at 0.297 stating that this regression explains 29.7 percent of variation in the data which we would still like to be higher but is a huge increase over our other models.  We can also see that about half of the team fixed effects are significant while the other half along with wpc_lag and relsal again are not significant.

####**Conclusion:**

After doing this lab and running all three regressions we can conclude that cricket, and the IPL in particular, seems to be the only league we studied where relsal and lagged win percentage both seem to have no significant effect on win percentage.  Being that our model is still not as good as we would hope for, since it only explains about 30 percent of variability in our data, a next step could be adding more different dependent variables along with some interactions to see if we can get a better model in terms of R-squared.  Doing this would also allow us to get a clearer picture to if relsal and lagged win percentage will ever be significant values in a regression or if they just don't matter as much as they do in other sports leagues to increase win percentage.
